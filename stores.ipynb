```python
# Vectorization: Vectorization means applying an operation to an entire array-like structure rather than manually processing each row with a Python loop.

# To define calculated variabled/derived variables
# Operating Cost+Total Customers*Acq Cost Per Cust
stores['TotalCost']=stores.OperatingCost + stores.Total_Customers*stores.AcqCostPercust
stores.head()
```

```python
# Calcuate 18% GST on Sales in 'GST' Column.
stores['GST']=0.18*stores['TotalSales']
stores.head()
```

```python
# assign: The assign() operation is used to create new columns while returning a DataFrame containing the new columns.
stores=stores.assign(TotalCost2=stores.OperatingCost+stores.Total_Customers*stores.AcqCostPercust,CP1=0.1)
stores.head()
```

```python
# insert- Not an efficient operation
stores.insert(2,'TotalCost3',stores.OperatingCost+stores.Total_Customers*stores.AcqCostPercust)
stores.head()
```

```python
# Task: Apply a 5% discount for superMarket Stores on Sales and 10% discount for everything else.
discount=[]
for index,row in stores.iterrows():
    if row['Store_Type']=='SuperMarket':
        discount.append(row['Sales']*0.05)
    else:
        discount.append(row['Sales']*0.1)
stores['discount']=discount
stores.head()            
```
