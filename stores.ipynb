```python
# If min is lower than Lower Cutoff(LC) then There is atleast one lower outliers.
# If max is greater than Upper Cutoff(UC) then There is atleast one upper outliers.
# Percentile Implementation
def percentile(x):
    """
    Apply the percentile rule on a series/column.
    Input:Series
    Output: lc,uc
    """
    lc=np.quantile(x,0.01)
    uc=x.quantile(0.99)
    print(f'min:{x.min()},lc:{lc},uc:{uc},max:{x.max()}')
    return lc,uc
lc,uc=percentile(stores.TotalSales)    
```

```python
# IQR Implementation
def iqr_rule(x):
    """
    Applying IQR rule on a Series/Column.
    Input: Series
    Output: lc,uc
    """
    q1=x.quantile(0.25)
    q3=x.quantile(0.75)
    iqr=q3-q1
    lc=q1-1.5*iqr
    uc=q3+1.5*iqr
    print(f"min:{x.min()},lc:{lc},uc:{uc},max:{x.max()}")
    return lc,uc
lc,uc=iqr_rule(stores.TotalSales)    
```

```python
# Z-score Implementation
def zscore_rule(x):
    """
    Applying z score rule on a series/column.
    Input: Series
    Output: lc,uc
    """
    mean=x.mean()
    SD=x.std()
    lc=mean-2.5*SD
    uc=mean+2.5*SD
    print(f"min:{x.min()},lc:{lc},uc:{uc},max:{x.max()}")
    return lc, uc
lc,uc=zscore_rule(stores.TotalSales)    
```
